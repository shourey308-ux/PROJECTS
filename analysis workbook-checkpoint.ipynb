```python
df1 = df1.drop_duplicates()
df2 = df2.drop_duplicates()
print("After duplicate - 2025:", df1.shape, "| 2026", df2.shape)
```

```python
df1 = df1[~(df1['Registration ID'].str.startswith('TEST-') | (df1['School'] == 'Test School'))]
df2 = df2[~(df2['Registration ID'].str.startswith('TEST-') | (df2['School'] == 'Test School'))]
print("After removing test rows — 2025:", df1.shape, "| 2026:", df2.shape)
```

```python
df1[df1['Registration ID'].str.startswith('TEST-')]
```

```python
def parse_date(s):
    s = str(s).strip()
    if '/' in s:
        return pd.to_datetime(s, format='%d/%m/%Y', errors='coerce')
    return pd.to_datetime(s, format='%Y-%m-%d', errors='coerce')

df1['Registration Date'] = df1['Registration Date'].apply(parse_date)
df2['Registration Date'] = df2['Registration Date'].apply(parse_date)
print("Any dates that failed to parse — 2025:", df1['Registration Date'].isna().sum(), "| 2026:", df2['Registration Date'].isna().sum())
```

```python
df1['Competition'] = df1['Competition'].str.strip().str.replace('Intl.', 'International', regex=False)
df2['Competition'] = df2['Competition'].str.strip().str.replace('Intl.', 'International', regex=False)

print(sorted(df1['Competition'].unique()))
```

```python
def clean_school(name):
    if pd.isna(name):
        return name
        n = str(name).strip().replace('Intl.', 'International').title()

df1['School'] = df1['School'].apply(clean_school)
df2['School'] = df2['School'].apply(clean_school)

print(sorted(df1['School'].dropna().unique()))
```

```python
for df in [df1, df2]:
    df['School'] = df['School'].fillna('Unknown School')
    df['City'] = df['City'].fillna('Unknown')
    df['Registration Source'] = df['Registration Source'].fillna('Unknown')

print("Remaining nulls — 2025:\n", df1.isnull().sum())
print("Remaining nulls — 2026:\n", df2.isnull().sum())
```

```python
df1.to_excel('cleaned_2025.xlsx', index=False)
df2.to_excel('cleaned_2026.xlsx', index=False)
print("Saved cleaned_2025.xlsx and cleaned_2026.xlsx")
```

## 2. Compare 2026 with 2025

```python
df1 = pd.read_excel('cleaned_2025.xlsx')
df2 = pd.read_excel('cleaned_2026.xlsx')
df1['Registration Date'] = pd.to_datetime(df1['Registration Date'])
df2['Registration Date'] = pd.to_datetime(df2['Registration Date'])
```

```python
cutoff = df2['Registration Date'].max()
print("Latest 2026 registration date:", cutoff)
```

```python
period_start = pd.Timestamp(year=2025, month=1, day=1)
period_end = pd.Timestamp(year=2025, month=cutoff.month, day=cutoff.day)

df1_period = df1[(df1['Registration Date'] >= period_start) & (df1['Registration Date'] <= period_end)]
print("2025 same-period rows:", len(df1_period), "| Full 2025 rows:", len(df1))
```

```python
summary = pd.DataFrame({
    '2025 (Jan 1 - Sep 4)': df1_period.groupby('Competition').size(),
    '2026 (Jan 1 - Sep 4)': df2.groupby('Competition').size()
}).fillna(0)

summary['Absolute Change'] = summary['2026 (Jan 1 - Sep 4)'] - summary['2025 (Jan 1 - Sep 4)']
summary['Pct Change (%)'] = (summary['Absolute Change'] / summary['2025 (Jan 1 - Sep 4)'] * 100).round(1)
summary['Status'] = summary['Pct Change (%)'].apply(lambda x: 'Growing' if x > 0 else ('Declining' if x < 0 else 'Flat'))
summary = summary.sort_values('Pct Change (%)', ascending=False)

summary
```

```python
summary.to_excel('yoy_comparison.xlsx')
print("Saved yoy_comparison.xlsx")
```

## 3. Analyze the trend

```python
df1_period = df1_period.copy()
df1_period['Month'] = df1_period['Registration Date'].dt.to_period('M')
df2['Month'] = df2['Registration Date'].dt.to_period('M')

monthly_1 = df1_period.groupby('Month').size()
monthly_2 = df2.groupby('Month').size()

print("2025 monthly:\n", monthly_1)
print("\n2026 monthly:\n", monthly_2)
```

```python
df2['Week'] = df2['Registration Date'].dt.to_period('W')
weekly_2 = df2.groupby('Week').size()
weekly_2.tail(10)
```

```python
last4_2 = df2[df2['Registration Date'] > cutoff - pd.Timedelta(weeks=4)]

last4_1_end = period_end
last4_1_start = period_end - pd.Timedelta(weeks=4)
last4_1 = df1_period[df1_period['Registration Date'] > last4_1_start]

last4_compare = pd.DataFrame({
    '2025_last4wk': last4_1.groupby('Competition').size(),
    '2026_last4wk': last4_2.groupby('Competition').size()
}).fillna(0)

last4_compare['Pct Change (%)'] = ((last4_compare['2026_last4wk'] - last4_compare['2025_last4wk']) 
                                     / last4_compare['2025_last4wk'] * 100).round(1)
last4_compare.sort_values('Pct Change (%)')
```

```python
monthly_2.to_frame('Registrations').to_excel('monthly_trend_2026.xlsx')
last4_compare.to_excel('last4weeks_comparison.xlsx')
print("Saved monthly_trend_2026.xlsx and last4weeks_comparison.xlsx")
```

```python
df1 = pd.read_excel('cleaned_2025.xlsx')
df2 = pd.read_excel('cleaned_2026.xlsx')
df1['Registration Date'] = pd.to_datetime(df1['Registration Date'])
df2['Registration Date'] = pd.to_datetime(df2['Registration Date'])
```

```python
cutoff = df2['Registration Date'].max()
period_start = pd.Timestamp(year=2025, month=1, day=1)
period_end = pd.Timestamp(year=2025, month=cutoff.month, day=cutoff.day)
df1_period = df1[(df1['Registration Date'] >= period_start) & (df1['Registration Date'] <= period_end)].copy()

summary = pd.DataFrame({
    '2025 (Same Period)': df1_period.groupby('Competition').size(),
    '2026': df2.groupby('Competition').size()
}).fillna(0).astype(int)
summary['Change'] = summary['2026'] - summary['2025 (Same Period)']
summary['Pct Change'] = (summary['Change'] / summary['2025 (Same Period)'] * 100).round(1)
summary = summary.sort_values('Pct Change', ascending=False).reset_index()
summary
```

```python
df2['Month'] = df2['Registration Date'].dt.to_period('M').astype(str)
monthly2 = df2.groupby('Month').size().reset_index(name='2026 Registrations')
monthly2
```

```python
df1['Year'] = 2025
df2['Year'] = 2026
combined = pd.concat([df1, df2], ignore_index=True)

with pd.ExcelWriter('GEMA_PowerBI_Data.xlsx', engine='openpyxl') as writer:
    combined.to_excel(writer, sheet_name='Cleaned_Data', index=False)
    summary.to_excel(writer, sheet_name='YoY_Summary', index=False)
    monthly2.to_excel(writer, sheet_name='Monthly_Trend_2026', index=False)
    last4_compare.reset_index().to_excel(writer, sheet_name='Last4Weeks', index=False)

print("Saved GEMA_PowerBI_Data.xlsx with 4 sheets")
```
